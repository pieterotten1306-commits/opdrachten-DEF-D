## Opdracht 2b: Opsplitsen taak 2: afleiding ontwerpeis

$| (1)/(1- (w)^2/(w_0)^2) | = 1/3$

met $w = 2\cdot \pi \cdot f$ en $f = 5 Hz$, opstelling massa $m = 0.5 kg$
$ C = 123 $

```python
spring_constant = 123
mass = 0.5
omega_0 = np.sqrt(spring_constant/mass)

frequencies = np.geomspace(0.1, 1000, 250)

def OmegaFormula(frequency):
    Omega = 2*np.pi*frequency
    Ratio = np.abs( (1) / (1-( (Omega**2) / (omega_0**2) )) )
    return Ratio

Ratios = np.zeros(len(frequencies))
for i in range(len(frequencies)):
    Ratios[i] = OmegaFormula(frequencies[i])



plt.figure()
plt.loglog(frequencies, Ratios)
plt.savefig("parameters.png")
plt.show()


```

## Opdracht 3: Plot de grafiek.

```python

def DeltaX(value, Mass):
    return (g*Mass)/value

values = (0.04, 0.08, 0.20)
Mass = np.linspace(0, 1, 100)

DeltaX1, DeltaX2, DeltaX5 = np.zeros(len(Mass)), np.zeros(len(Mass)), np.zeros(len(Mass))

for i in range(len(Mass)):
    DeltaX1[i] = DeltaX(values[0], Mass[i])
    DeltaX2[i] = DeltaX(values[1], Mass[i])
    DeltaX5[i] = DeltaX(values[2], Mass[i])

plt.plot(Mass, DeltaX1, label='1 DeltaX')
plt.plot(Mass, DeltaX2, label='2 DeltaX')
plt.plot(Mass, DeltaX5, label='5 DeltaX')

plt.xlabel('Veer constant (C)')
plt.ylabel('Massa (kg)')
plt.savefig('DeltaX.png')
plt.show()
plt
```

```python
spring_constant_1 = 123
mass_1 = 0.5
omega_0_1 = np.sqrt(spring_constant_1/mass_1)


spring_constant_2 = 40
mass_2 = 0.3
omega_0_2 = np.sqrt(spring_constant_2/mass_2)

frequencies = np.geomspace(0.5, 10, 250)

def OmegaFormula(frequency, omega_0):
    Omega = 2*np.pi*frequency
    Ratio = np.abs( (1) / (1-( (Omega**2) / (omega_0**2) )) )
    return Ratio

Ratios_1 = np.zeros(len(frequencies))
for i in range(len(frequencies)):
    Ratios_1[i] = OmegaFormula(frequencies[i], omega_0_1)

Ratios_2 = np.zeros(len(frequencies))
for i in range(len(frequencies)):
    Ratios_2[i] = OmegaFormula(frequencies[i], omega_0_2)

plt.figure()
plt.loglog(frequencies, Ratios_1)
plt.loglog(frequencies, Ratios_2)
plt.xlabel('Frequentie (Hz)')
plt.ylabel('Amplitude-overdracht')

plt.show()
```
